```python
import os
import pandas as pd

# Define the base path for the 2025 operational daily rain data
base_path = '/content/drive/MyDrive/midtermdata/BSC DPDM midterm 2025/Input - (Operation Daily rain from API_ 2025)'

# Create a dictionary to store DataFrames for each month
monthly_dfs = {}

# Loop through months from 7 (July) to 12 (December) for 2025
for month in range(7, 13):
    month_str = f'2025-{month:02d}' # Format month as '07', '08', etc.
    month_path = os.path.join(base_path, month_str)

    # Check if the directory exists
    if os.path.exists(month_path):
        all_daily_files = []
        # List all CSV files in the current month's directory
        for file_name in os.listdir(month_path):
            if file_name.endswith('.csv'):
                file_path = os.path.join(month_path, file_name)
                # Read each daily CSV file and append to a list
                all_daily_files.append(pd.read_csv(file_path))

        # Concatenate all daily DataFrames for the current month into one DataFrame
        if all_daily_files:
            monthly_dfs[f'df_2025_{month:02d}'] = pd.concat(all_daily_files, ignore_index=True)
            print(f"Loaded {len(all_daily_files)} daily files for {month_str} into df_2025_{month:02d}")
        else:
            print(f"No CSV files found in {month_path}")
    else:
        print(f"Directory not found: {month_path}")

# You can now access each monthly DataFrame, e.g., monthly_dfs['df_2025_07']
# To verify, print the first few rows of the July DataFrame if it exists
if 'df_2025_07' in monthly_dfs:
    print("\nFirst 5 rows of df_2025_07:")
    print(monthly_dfs['df_2025_07'].head())
```

#2. กรองข้อมูลให้เหลือเฉพาะสถานีใน stations in the model_hii.csv

```python
# สร้าง list ของสถานีที่ต้องการ (Target Stations)
target_stations = stations_in_the_modle_hii['Node'].unique()

print(f"จำนวนสถานีที่มีในข้อมูลดิบ (2025): {all_2025_daily_rain['tele_station_oldcode'].nunique()} สถานี")

# กรองข้อมูล: เลือกเฉพาะ row ที่ column 'tele_station_oldcode' อยู่ใน target_stations
df_2025_filtered = all_2025_daily_rain[all_2025_daily_rain['tele_station_oldcode'].isin(target_stations)].copy()

print(f"จำนวนสถานีหลังจากกรองเหลือ: {df_2025_filtered['tele_station_oldcode'].nunique()} สถานี")
print(f"จำนวน Row ก่อนกรอง: {len(all_2025_daily_rain)} -> หลังกรอง: {len(df_2025_filtered)}")
```

#3. ระบุข้อมูลสูญหาย (Missing Records) ของแต่ละสถานี

```python
# สร้างช่วงเวลาเต็ม ตั้งแต่ 1 ก.ค. - 31 ธ.ค. 2025
full_date_range = pd.date_range(start='2025-07-01', end='2025-12-31', freq='D')

# แปลงคอลัมน์วันที่ใน df_2025_filtered ให้เป็น datetime object
df_2025_filtered['measure_datetime'] = pd.to_datetime(df_2025_filtered['measure_datetime'])

# แปลงข้อมูลเป็น Pivot Table เพื่อให้เห็นช่องว่างง่ายๆ (Index=Date, Column=Station)
# ถ้าวันไหนไม่มีข้อมูล ค่าจะเป็น NaN โดยอัตโนมัติ
df_pivot = df_2025_filtered.pivot_table(index='measure_datetime', columns='tele_station_oldcode', values='rainfall_value')

# Reindex ด้วยช่วงเวลาเต็ม เพื่อให้เห็นวันที่หายไปทั้งแถว (ถ้ามี)
df_pivot = df_pivot.reindex(full_date_range)

# นับจำนวนข้อมูลที่หายไป (NaN) ในแต่ละสถานี
missing_count = df_pivot.isnull().sum()
total_days = len(full_date_range) # 184 วัน

# คำนวณสัดส่วน % ที่หายไป
missing_percent = (missing_count / total_days) * 100

# สร้าง Dataframe สรุปผล
missing_summary = pd.DataFrame({
    'Missing_Records': missing_count,
    'Percent_Missing': missing_percent
})

print("=== สรุปข้อมูลสูญหายแต่ละสถานี (ตัวอย่าง 10 สถานีแรก) ===")
print(missing_summary.head(10))
```

#4. คัดสถานีที่ข้อมูลสูญหาย >= 20% ออก

```python
# หา list รายชื่อสถานีที่ต้องตัดทิ้ง (Missing >= 20%)
stations_to_drop = missing_summary[missing_summary['Percent_Missing'] >= 20].index.tolist()

# หา list รายชื่อสถานีที่จะเก็บไว้ (Missing < 20%)
stations_to_keep = missing_summary[missing_summary['Percent_Missing'] < 20].index.tolist()

print(f"จำนวนสถานีทั้งหมด: {len(missing_summary)}")
print(f"ถูกคัดออก (>=20% missing): {len(stations_to_drop)} สถานี")
print(f"คงเหลือ: {len(stations_to_keep)} สถานี")

print("\nรายชื่อสถานีที่ถูกคัดออก:", stations_to_drop)

# กรองข้อมูล Pivot Table ให้เหลือเฉพาะสถานีที่เก็บไว้
df_clean_pivot = df_pivot[stations_to_keep].copy()
```

#5. เติมข้อมูลที่หายไปด้วยค่า Median ตาม Day of Year จาก Historical

```python
# --- 5.1 เตรียมข้อมูล Historical ---
# แปลง measure_datetime เป็น datetime
hii_daily_rain['measure_datetime'] = pd.to_datetime(hii_daily_rain['measure_datetime'])
# สร้าง column Day of Year (1-366)
hii_daily_rain['DOY'] = hii_daily_rain['measure_datetime'].dt.dayofyear

# คำนวณ Median แยกตาม 'station_code' และ 'DOY'
historical_median = hii_daily_rain.groupby(['station_code', 'DOY'])['data'].median().reset_index()
historical_median.rename(columns={'data': 'median_rain', 'station_code': 'code'}, inplace=True)

# --- 5.2 เติมข้อมูลลงใน df_clean_pivot ---
# แปลง Pivot กลับมาเป็นตารางยาว (Melt/Stack) เพื่อให้จัดการง่าย
df_filled = df_clean_pivot.stack(dropna=False).reset_index()
# หลังจาก stack และ reset_index คอลัมน์จะเป็น 'measure_datetime', 'tele_station_oldcode' และ 'rainfall_value'
# เราจะเปลี่ยนชื่อคอลัมน์เหล่านี้เพื่อให้ตรงกับที่เราต้องการ ('date', 'code', 'rain')
df_filled.columns = ['date', 'code', 'rain']

# เพิ่ม column DOY ให้ข้อมูลปี 2025
df_filled['DOY'] = df_filled['date'].dt.dayofyear

# Merge ข้อมูล Median เข้ามา (เหมือน VLOOKUP)
df_filled = pd.merge(df_filled, historical_median, on=['code', 'DOY'], how='left')

# เติมค่า NaN ใน column 'rain' ด้วยค่าจาก 'median_rain'
df_filled['rain'] = df_filled['rain'].fillna(df_filled['median_rain'])

# ตรวจสอบว่ายังมีค่าหายไปไหม
missing_after_fill = df_filled['rain'].isnull().sum()
print(f"จำนวน Record ที่ยังเป็น NaN หลังเติมข้อมูล: {missing_after_fill}")

# (Optional) ถ้ายังมี NaN (เพราะ Historical ก็ไม่มีข้อมูลวันนั้น) อาจเติมด้วย 0 หรือค่าเฉลี่ยรวม
if missing_after_fill > 0:
    df_filled['rain'] = df_filled['rain'].fillna(0)
    print("เติม NaN ที่เหลือด้วย 0 เรียบร้อย")
```

#6. บอกเหตุผลที่ใช้ Median แทน Mean

**เหตุผลที่ใช้ Median (มัธยฐาน) แทน Mean (ค่าเฉลี่ย):**

1.  **การแจกแจงข้อมูลน้ำฝน (Skewed Distribution):** ข้อมูลปริมาณน้ำฝนส่วนใหญ่มักเป็น 0 ในวันที่ฝนไม่ตก และมีบางวันที่ค่าสูงมาก ในวันที่พายุเข้า ทำให้กราฟเบ้ขวาอย่างมาก
2.  **ผลกระทบจากค่าผิดปกติ (Outliers):** ค่า Mean จะถูกดึงด้วยค่าฝนที่ตกหนักผิดปกติ (Extreme values) ทำให้ค่าที่ได้สูงเกินความเป็นจริงของวันปกติ
3.  **ตัวแทนที่ดีกว่า:** Median คือค่าที่อยู่ตรงกลางจริงๆ จึงเป็นตัวแทนที่เหมาะสมกว่าในการระบุว่าโดยปกติแล้ว ในวันนี้ของปี ฝนมักจะตกประมาณเท่าไหร่

#7. คำนวณผลรวมปริมาณน้ำฝนรายเดือน และจัดรูปแบบ Output

```python
# สร้าง column month และ year จาก date
df_filled['month'] = df_filled['date'].dt.month
df_filled['year'] = df_filled['date'].dt.year

# คำนวณผลรวมรายเดือน (Groupby Station, Month, Year)
monthly_rain = df_filled.groupby(['code', 'month', 'year'])['rain'].sum().reset_index()

# เปลี่ยนชื่อ column 'code' เป็น 'Node' ให้ตรงตามโจทย์
monthly_rain.rename(columns={'code': 'Node'}, inplace=True)

# จัดลำดับ Column ให้ตรงตามไฟล์ตัวอย่าง (Node, rain, month, year)
output_df = monthly_rain[['Node', 'rain', 'month', 'year']]

# แสดงผลลัพธ์
print("=== ตัวอย่างผลลัพธ์รายเดือน (5 แถวแรก) ===")
print(output_df.head())

# บันทึกไฟล์ (Optional)
# output_df.to_csv('output_monthly_rainfall.csv', index=False)
```
